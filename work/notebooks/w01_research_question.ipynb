
## 1. Lane 3: Structured Content Archetype Clustering using the starter dataset.

I want to focus on low-performing pages and explore whether their search performance signals form distinct recurring groups. M main aim is to investigate whether differences in metrics such as impressions, CTR, and average position reveal useful archetypes that could support different review or optimization actions. I chose this lane because clustering may help uncover patterns without requiring predefined labels, while still keeping the result exploratory rather than treating the clusters as objectively true page types. ## 2. The question: decision, action, cost of a wrong call

**Research question:** Can low-performing pages be grouped into distinct search performance archetypes using impressions, CTR, and average position, so content teams can choose a more appropriate optimization action?

**Unit of analysis:** one page/content item. **Output:** a cluster assignment and a profile describing each recurring performance pattern.

The decision is how a content or SEO team should respond to an underperforming page. A simple rule such as “zero clicks = poor performance” may group together pages that are underperforming in different ways, while clustering can combine several observed signals to explore recurring patterns. A wrong recommendation could waste time on the wrong pages or lead to the wrong intervention, so the clusters should support investigation rather than make decisions automatically.


## 3. Quick look at the data



```python
import pandas as pd

# Load the starter dataset from the repository
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Count the total number of pages in the dataset.
# Each row represents one content page.
total_pages = len(df)

# Count how many pages received no clicks in the last 90 days.
# This helps us see how common low click performance is.
zero_clicks = (df["clicks_90d"] == 0).sum()

# Convert the number of zero-click pages into a percentage
# so it is easier to compare with the whole dataset.
zero_clicks_pct = zero_clicks / total_pages * 100

# Look for pages that had some search visibility but still received no clicks.
# I use 1,000 impressions here only as a simple exploratory threshold,
# not as a definition of what "high impressions" means.
high_impressions_zero_clicks = (
    (df["impressions_90d"] >= 1000)
    & (df["clicks_90d"] == 0)
).sum()

# Print the numbers that support the research question.
print(f"Total pages: {total_pages:,}")
print(f"Pages with zero clicks: {zero_clicks:,} ({zero_clicks_pct:.1f}%)")
print(
    f"Pages with at least 1,000 impressions but zero clicks: "
    f"{high_impressions_zero_clicks:,}"
)
```

The dataset contains 30,000 pages, and 13,204 pages (44.0%) received zero clicks in the last 90 days. However, 1,149 pages had at least 1,000 impressions and still received zero clicks. This suggests that pages with the same outcome, such as zero clicks, may be underperforming for different reasons. Some may have low visibility, while others are visible in search but do not attract clicks. This supports exploring whether clustering can reveal different search performance archetypes among underperforming pages.


## 4. Careful words: what I can and can't claim

This analysis can show whether pages with similar search-performance metrics tend to form recurring groups in the starter dataset. These clusters may help describe different patterns of underperformance and support further investigation.

However, the clusters are not true labels, and they do not prove why a page is underperforming or what action will definitely improve it.
