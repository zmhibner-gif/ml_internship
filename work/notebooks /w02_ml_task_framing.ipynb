## 1. My lane as an ML task (type)

My chosen lane is Structured Content Archetype Clustering, so the task type is clustering.

I want to explore whether pages with similar search-performance signals, such as impressions, CTR, and average position, form distinct patterns. Clustering fits this because I am not starting with a specific outcome in mind. Instead, I want to see what patterns appear in the data and then interpret what they mean.

I chose this lane partly because I already have some experience with clustering from university, but I also like its exploratory nature. I do not know beforehand what groups the data will form, which i find very interesting,  and I want to understand this type of analysis better through the project.

## 2. Target or proxy

Because this is a clustering task, there is no target variable that I am trying to predict.

Instead, the model will assign each page to a cluster based on similarities in search-performance features such as impressions, CTR, and average position. The resulting cluster assignment would be an output of the analysis rather than an existing label in the dataset.

These clusters should be treated as exploratory patterns, not as true page categories or explanations for why a page is underperforming.

## 3. Success metric

Since this is a clustering task, I cannot use metrics such as accuracy because there are no correct cluster labels to compare against.

I will use the silhouette score as the main quantitative metric. It measures how similar pages are to others in the same cluster compared with pages in other clusters. A higher score generally means the clusters are more clearly separated.

I will also inspect the clusters themselves to check whether they represent understandable search-performance patterns that could be useful for content decisions.

## 4. The unit of analysis, as a real dataframe

The unit of analysis is one content page. Each row represents one anonymized page and contains its search-performance measurements from the previous 90 days.

For my clustering lane, I am currently focusing on impressions, CTR, and average position. I keep content_id in the dataframe only to identify each page; it will not be used as a clustering feature.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv") # Load the
#anonymized starter dataset

lane_df = df[["content_id", "impressions_90d", "ctr", "avg_position"]].copy()
# Keep only the columns needed for my clustering lane
lane_df = lane_df[lane_df["avg_position"] > 0]
# avg_position = 0 means position data is unavailable

print(lane_df.shape[0], "rows,", lane_df.shape[1], "columns")
lane_df.head(10)

#For this initial framing, I am focusing on impressions, CTR, and average
#position. I may include additional variables later if they add meaningful
#information to the clustering, but I would first justify their relevance and
#check how they affect the results.
```

## 5. Why ML beats a fixed rule here

A fixed rule would require me to decide thresholds in advance, for example what counts as high impressions or low CTR. This could group together pages that have very different performance problems.

Clustering allows me to look at several performance signals together and see what patterns appear in the data instead of defining all the groups myself. This could reveal different types of underperformance that may need different content actions. For example, different clusters could help a content team decide which pages should be monitored, improved, rewritten, or reviewed more closely.

However, clustering is only useful if the resulting groups are clear and meaningful. If they are not, a simpler rule-based approach may be more appropriate.
